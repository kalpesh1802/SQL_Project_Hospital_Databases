## Generate 3000 Patients

```python
from faker import Faker
import pandas as pd
import random

fake = Faker('en_IN')

patients = []

for i in range(3000):
    patients.append([
        fake.name(),
        random.randint(1, 90),
        random.choice(['Male', 'Female', 'Other']),
        random.choice(['A+', 'A-', 'B+', 'B-', 'AB+', 'AB-', 'O+', 'O-']),
        fake.city(),
        fake.msisdn()[:10]
    ])

df = pd.DataFrame(
    patients,
    columns=[
        'name',
        'age',
        'gender',
        'blood_group',
        'city',
        'contact'
    ]
)

df.to_csv('patients.csv', index=False)

print("3000 patient records generated.")
```

## Generate 200 Doctors

```python
from faker import Faker
import pandas as pd
import random

fake = Faker('en_IN')

specializations = [
    'Cardiologist',
    'Neurologist',
    'Orthopedic Surgeon',
    'Pediatrician',
    'Dermatologist',
    'Oncologist',
    'Psychiatrist',
    'General Physician'
]

doctors = []

for i in range(200):
    doctors.append([
        fake.name(),
        random.choice(specializations),
        random.randint(1, 10),
        random.randint(1, 30)
    ])

df = pd.DataFrame(
    doctors,
    columns=[
        'doctor_name',
        'specialization',
        'dept_id',
        'experience_yrs'
    ]
)

df.to_csv('doctors.csv', index=False)
print("200 Doctors Record Genrated")
```

## Generate 10,000 Admissions

```python
from faker import Faker
import pandas as pd
import random

fake = Faker()

admissions = []

for i in range(10000):
    admissions.append([
        random.randint(1, 3000),
        random.randint(1, 200),
        fake.date_between(
            start_date='-2y',
            end_date='today'
        ),
        fake.date_between(
            start_date='today',
            end_date='+30d'
        ),
        random.choice([
            'General',
            'ICU',
            'Private',
            'Emergency'
        ]),
        random.choice([
            'Admitted',
            'Discharged',
            'Under Observation'
        ])
    ])

df = pd.DataFrame(
    admissions,
    columns=[
        'patient_id',
        'doctor_id',
        'admit_date',
        'discharge_date',
        'ward_type',
        'status'
    ]
)

df.to_csv('admissions.csv', index=False)
print("10,000 Admission Record Genrated")
```

## Generate 10,000 Diagnoses

```python
from faker import Faker
import pandas as pd
import random

diagnoses_list = [
    ('I21', 'Heart Attack'),
    ('J18', 'Pneumonia'),
    ('I10', 'Hypertension'),
    ('E11', 'Diabetes'),
    ('G43', 'Migraine'),
    ('M17', 'Osteoarthritis')
]

records = []

for i in range(10000):
    code, disease = random.choice(diagnoses_list)

    records.append([
        random.randint(1, 10000),
        code,
        disease,
        random.choice([
            'Mild',
            'Moderate',
            'Severe',
            'Critical'
        ])
    ])

df = pd.DataFrame(
    records,
    columns=[
        'admission_id',
        'icd_code',
        'diagnosis_name',
        'severity'
    ]
)

df.to_csv('diagnoses.csv', index=False)
print("10,000 Diagnoses record genrated")
```

## Generate 10,000 Billing Records

```python
from faker import Faker
import pandas as pd
import random

billing = []

for i in range(10000):

    total = random.randint(5000, 500000)

    insurance = random.randint(
        0,
        int(total * 0.8)
    )

    billing.append([
        random.randint(1, 10000),
        total,
        insurance,
        random.choice([
            'Paid',
            'Pending',
            'Waived'
        ]),
        random.choice([
            'Cash',
            'Card',
            'Insurance',
            'UPI'
        ])
    ])

df = pd.DataFrame(
    billing,
    columns=[
        'admission_id',
        'total_amount',
        'insurance_covered',
        'payment_status',
        'payment_mode'
    ]
)

df.to_csv('billing.csv', index=False)
print("10,000 Billing Records Genrated")
```
